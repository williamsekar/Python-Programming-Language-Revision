```python
# IF
age = 3
if age<=12:
    print("He/She is a child!")
```

```python
# NESTED-IF
age = 15
gender = "male"
if gender == "male":
    if age<=19:
        print("He is a teen!")
```

```python
# IF-ELSE
age = 20
if age>=20 and age<=35:
    print("He/She is Adult")
else:
    print("He/She is not Adult")
```

```python
# NESTED IF-ELSE
age = 40
if age<=12:
    print("Child")
elif age>=13 and age<=19: 
    print("Teen")
elif age>=20 and age<=35:
    print("Adult")
else:
    print("old")
```

```python
# MATCH-CASE STATEMENT
day = 3
match day:
    case 1:
        print("Mon")
    case 2:
        print("Tue")
    case 3:
        print("Wed")
    case 4:
        print("Thu")
    case 5:
        print("Fri")
    case 6:
        print("Sat")
    case 7:
        print("Sun")
    case _:
        print("Wrong Input")    
```
