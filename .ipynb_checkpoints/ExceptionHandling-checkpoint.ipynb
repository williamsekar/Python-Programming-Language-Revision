```python
# TRY / EXCEPT
```

```python
try:
    a=10
    b=0
    result= a/b
    print(result)
except:
    print("Someting wrong")
```

```python
try:
    a=10
    b=0
    result= a/b
    print(result)
except ZeroDivisionError:
    print("You can't divide by zero")
```

```python
## MULTIPLE EXCEPT
```

```python
try:
    number=int(input("Enter the number"))
    result=100/number
    print(result)
except ValueError:
    print("Please enter the number")
except ZeroDivisionError:
    print("You can't divide by zero")
```

```python
# FINALLY
```

```python
try:
    number=int(input("Enter the number"))
    result=100/number
    print(result)
except ValueError:
    print("Please enter the number")
except ZeroDivisionError:
    print("You can't divide by zero")
finally:
    print("Program is finished!")
```

```python
# CUSTOM ERROR
```

```python
class InsuffientAmountError(Exception):
    pass
try:
    balance= 5000
    withdraw= 10000
    if withdraw > balance:
        raise InsuffientAmountError("Insuffient amount!")
    balance=-withdraw
    print("Amount debited!")
except InsuffientAmountError as e:
    print(e)
```

```python
# CHANGE THE INBULIT ERROR
```

```python
class ValueError(Exception):
    pass

def set_age(age):
    if age < 0:
        raise ValueError("Age must be postive value")
    return age
try:
    age= int(input("Enter the age: "))
    result=set_age(age)
    print(result)
except ValueError as e:
    print(e)
    
```

```python
## USING EXCEPTION TO FILE HANDLING 
```

```python
file =None
try:
    file = open("data.txt","r")
    data= file.read()
    print(data)
except FileNotFoundError:
    print("File not found!")
finally:
    if file:
        file.close()
```
