```python
# FILE HANDLING
```

```python
# READING A FILE
file= open("data.txt","r")
content=file.read()
print(content)
file.close()
```

```python
# USING WITH TO AVOID FILE CLOSE
```

```python
with open("data.txt","r") as file:
    content=file.read()
    print(content)
```

```python
# FILE READ LINE BY LINE
```

```python
with open("data.txt","r") as file:
    print(file.readline())
    print(file.readline())
    print(file.readline())
```

```python
# FILE READING MULTIPLE LINES
```

```python
with open("data.txt","r") as file:
    print(file.readlines()) # return as line with escape characters
```

```python
# CSV - COMMAS SEPARATED VALUES
'''
name,age,gender
william,28,male
sekar,22,male
prema,18,female
'''
```

```python
# READ A CSV FILE AS LIST
import csv

with open("students.csv","r") as file:
    data=csv.reader(file)
    for item in data:
        print(item)
```

```python
# READ A CSV FILE AS DICTIONARY
import csv

with open("students.csv","r") as file:
    data=csv.DictReader(file)
    for item in data:
        print(item)
        print(item["name"])
```

```python
# WRITE A CSV FILE  AS LIST
```

```python
import csv
data=[["name","color"],["apple","red"],["orange","orange"],["banana","yellow"]]
with open("fruits.csv","w",newline="\n") as file:
    writer=csv.writer(file)
    writer.writerows(data)
```

```python
# WRITE A CSV FILE AS DICTIONARY
```

```python
import csv
data=[ {"species_name":"lion","leg":4},{"species_name":"parrot","leg":2},{"species_name":"snake","leg":0} ]
with open("animal.csv","w",newline="\n") as file:
    fieldnames=["species_name","leg"]
    writer=csv.DictWriter(file,fieldnames)
    writer.writeheader()
    writer.writerows(data)
```

```python
# JSON
```

```python
## PYTHON OBJECT TO JSON STRING
import json
python_object=[{"name":"william", "age":28},
               {"name":"perame", "age":22}, 
               {"name":"mani","age":23}
              ]
json_data = json.dumps(python_object,indent=2)
print(json_data)    
print(type(json_data))

## JSON STRING TO PYTHON OBJECT
result= json.loads(json_data)
print(result)
print(type(result))

## WRITE JSON TO FILE
with open("students.json","w") as file:
    json.dump(python_object,file,indent=2)
    
## READING JSON TO FILE
with open("students.json","r") as file:
    result = json.load(file)
    print(result)
```
