```python
# WHILE LOOP
i=1
while i<=5:
    print(i)
    i+=1
```

```python
# WHILE WITH BREAK
j=1
while j<=10:
    if(j==4):
        break
    print(j)
    j+=1
```

```python
# WHILE WITH CONTINUE
k=1
while k<=5:
    if(k==3):
        k+=1
        continue       
    print(k)
    k+=1
```

```python
# WHILE ELSE
i=1
while i<=5:
    print(i)
    i+=1
else:
    print("Loop executed")
    
```

```python
# NESTED WHILE LOOP

i=0
row=2
col=2
while i<=row:
    j=0
    while j<=col:
        step=1
        if step<=3:
            print("[",i,"]","[",j,"]", end="\t")
        step+=1
        j+=1
    print("",end="\n")
    i+=1
        
```

```python
# FOR LOOP - RANGE(END)
for i in range(5):
    print(i)
    i+=1
```

```python
# FOR LOOP - RANGE(START, END)
for i in range(5,11):
    print(i)
    i+=1
```

```python
# FOR LOOP - RANGE(START, END, STEP)
for i in range(100,111,2):
    print(i)
    i+=1
```

```python
# FOR LOOP - REVERSE
for i in range(5,0,-1):
    print(i)
    i+=1
```

```python
# FOR LOOP WITH BREAK
for i in range(5):
    if i==3:
        break
    i+=1
    print(i)
```

```python
# FOR LOOP WITH CONTINUE
for i in range(1,6):
    if i==3:
        print("Skip")
        i+=1
        continue
    print(i)
    i+=1
```

```python
# FOR LOOP WITH ELSE
for i in range(6):
    i+=1
    print(i)
else:
    print("Loop executed!")
```
