```python
# 2. ASSIGNMENT OPERATORS
```

```python
## ASSIGN 
x = 5
print(x)
```

```python
# ADDITION & ASSIGN
```

```python
x += 3
print(x)
```

```python
## SUBSTRACTION & ASSIGN
x -= 3
print(x)
```

```python
## MULTIPLICATION & ASSIGN
x *= 3
print(x)
```

```python
## DIVISION & ASSIGN
x /= 3
print(x)
```

```python
## FLOOR DIVISION AND ASSIGN
x = 16
x //= 3
print(x)
```

```python
## MODULUS AND ASSIGN 
x %= 2
print(x)
```

```python
## POWER AND ASSIGN
x = 3
x **= 2
print(x)
```

```python
# 4. LOGICAL OPERATORS
```

```python
## and
print( (5==5) and (10>3) )
```

```python
## or
print( (7<=10) or (11!=11) )
```

```python
## not
print( not(3<=3) )
```

```python
# 5. UNARY OPERATORS
''' One operand '''
```

```python
## UNARY PLUS(+)
x = 10
print(+x)
print(-x)
```

```python
## UNARY MINUS(-)
y = 10
print(-y)
y = -10
print(-y)
```

```python
## UNARY LOGICAL OPERATOR
is_active = True
print(not is_active)
```

```python
# BITWISE OPEARATORS
```

```python
## BITWISE AND(&)
i = 5
j = 3
print(i & j)
```

```python
## BITWISE OR(|)
print(i | j)
```

```python
## BITWISE NOT(~)
print(~i)
```

```python
## BITWISE NAND(^)
print(i ^ j)
```

```python
## LEFT SHIFT(<<)
print(5<<2)
```

```python
## RIGHT SHIFT(>>)
print(20>>2)
```

```python
# 7. IDENTITY OPERATORS
```

```python
## IS OPERATOR
x = "Hello"
y = "Hello"
print(id(x))
print(id(y))
print(x is y) # python reuses a same memory space for small integers 
s1 = "Hello World"
s2 = "Hello World"
print(s1 is s2) # s1 and s2 variable uses same object but space occurs to give a result "False"
```

```python
## IS NOT OPERATOR
t1 = "Alice"
t2 = "Alice"
print(t1 is not t2)
u1 = "Alice Bob"
u2 = "Alice Bob"
print(u1 is not u2)
```
