```python
## CREATING A LIST
numbers=[10,20,30,40,50]
names=["Alice","Bob","Charlie"]
mixed=[10,"Alice",3.14,True]

## ACCESSING A LIST
print(numbers[0])
print(numbers[2])
print(numbers[-1])

## CHANGING ELEMENTS WITH POSITION
numbers[2]=200
print(numbers)

# IMPORATNT METHODS
print(len(numbers)) # length of list
new_numbers= numbers.copy() # Returns a shallow copy of list
print(new_numbers)
print(numbers is new_numbers)
numbers.append(600) # Append the value at end
print(numbers)
numbers.extend([10,40]) # Adds all elements at end
print(numbers)
numbers.insert(3,[True,False,True]) # Insert the value or list at position
print(numbers)
numbers.remove(20) # Remove a value
print(numbers)
numbers.pop() # Remove last item
print(numbers)
numbers.pop(2) # Remove item at index
print(numbers)
occurance=numbers.count(10) # Returns a occurance of value
print(occurance)
numbers.sort() # Sort
print(numbers)
numbers.reverse() # Reverse
print(numbers)
numbers.clear() # Remove everything
print(numbers)
```

```python
# DICTIONARY
```

```python
# CREATE A DICTIONARY
student={
    "name":"williams",
    "age": 25,
    "city": "chennai"
}
print(student)

# ACCESSING VALUES
print(student["name"]) # Get the value by key

# ADDING A NEW ITEM
student["gender"]= "male" # Insert the value by key and value
print(student)

# CHANGING A VALUE
student["age"]=29 # Change the value by key
print(student)

# REMOVING ITEMS
del student["city"] # Remove
print(student)

# POP()
age=student.pop("age") # It returns value of key and remove that
print(age)
print(student)

# GET()
print(student.get("name")) # Instead of student["Key"]

# NONE
print(student.get("age")) # Its return None, dictionary doesn' have the key 

# DEFAULT VALUE
print(student.get("email","Email not available"))

# CHECKING WHETHER A KEY EXISTS
print("name" in student)
print("age" in student)

# DICTIONARY LENGTH
print(len(student))

# GETTING ALL KEYS
print(student.keys())
print(list(student.keys()))

# GETTING ALL VALUES
print(student.values())
print(list(student.values()))

# GETTING KEYS AND VALUES TOGETHER
print(student.items())

# LOOP THROUGH A DICTIONARY
for key in student:
    print(key)
for value in student.values():
    print(value)
for key,value in student.items():
    print(key,"=",value)

# NESTED DICTIONARY
students={
    "student1":
    {
        "name":"Alice",
        "age": 25
    },
    "student2":
    {
        "name":"John",
        "age": 24
    }
}
print(students["student1"]["age"])

# DICTIONARY WITH LIST
student={
    "name" : "William",
    "Skils" : ['Python','C++','Java']
}
print(student["Skils"][1])

# COPY
new_student= student.copy()
print(new_student)

# FROMKEYS()
keys= ["name","age","city"]
s1=dict.fromkeys(keys) # New dictionary using a sequence of keys with None value
s2=dict.fromkeys(keys,"NULL")
print(s1)
print(s2)

# POPITEM()
item= s1.popitem() # Return value is tuple, last item is removed
print(s1)

# UPDATE
student={
    "name": "William",
    "age" : 25
}

student.update({
    "age": 26, # Change the existing value
    "city": "Chennnai" # Add new value
})
print(student)

# SETDEFAULT()
result= student.setdefault("gender", "Male")
print(result)
```

```python
# CREATE A SET
## USING {}
fruits = {"apple", "banana", "orange"} # Doesn't allow duplicate values
print(fruits)

## USING SET()
numbers = set([10, 20, 30, 40])
print(numbers)

## CREATE A EMPTY SET
x={} # it's dictionary
print(type(x)) 
x=set() # empty set
print(type(x))

# ADDING ITEMS
numbers = {10, 20, 30}
numbers.add(40)
print(numbers)

# REMOVE ITEMS
numbers = {10, 20, 30, 40}
numbers.remove(30)
print(numbers)

# REMOVE ITEMS WITHOUT ERROR
numbers = {10, 20, 30}
numbers.discard(50)
print(numbers)

# POP
numbers = {10, 20, 30, 40}
x = numbers.pop() # random remove
print(x)
print(numbers)

# CLEAR
numbers = {10, 20, 30}
numbers.clear()
print(numbers)

# DEL -SHOW THE NAMING ERROR
numbers = {10, 20, 30}
del numbers

# LENGTH
numbers = {10, 20, 30, 40}
print(len(numbers))

# MOST IMPORTANT SET OPERATIONS
## UNIOIN | SYMBOL
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.union(B))

## INTERSECTION & SYMBOL
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A.intersection(B))

## DIFFERENCE  -
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A - B) # exist in the first set

## SYMMENTIC DIFFERENCE ^
A = {1, 2, 3, 4} 
B = {3, 4, 5, 6}
print(A ^ B) # Elements that are in either set, but NOT in both

## SUBSET <=
A = {1, 2}
B = {1, 2, 3, 4}
print(A.issubset(B)) # if all of its elements exist inside another set

## SUPERSET  >=
A = {1, 2, 3, 4}
B = {1, 2}
print(A.issuperset(B))

## DISJOINT SETS
A = {1, 2, 3}
B = {4, 5, 6}
print(A.isdisjoint(B)) # no common elements
```
