```python
## CREATING A LIST
numbers=[10,20,30,40,50]
names=["Alice","Bob","Charlie"]
mixed=[10,"Alice",3.14,True]

## ACCESSING A LIST
print(numbers[0])
print(numbers[2])
print(numbers[-1])

## CHANGING ELEMENTS WITH POSITION
numbers[2]=200
print(numbers)

# IMPORATNT METHODS
print(len(numbers)) # length of list
new_numbers= numbers.copy() # Returns a shallow copy of list
print(new_numbers)
numbers.append(600) # Append the value at end
print(numbers)
numbers.extend([10,40]) # Adds all elements at end
print(numbers)
numbers.insert(3,77) # Insert the value or list at position
print(numbers)
numbers.remove(20) # Remove a value
print(numbers)
numbers.pop() # Remove last item
print(numbers)
numbers.pop(2) # Remove item at index
print(numbers)
occurance=numbers.count(10) # Returns a occurance of value
print(occurance)
numbers.sort() # Sort
print(numbers)
numbers.reverse() # Reverse
print(numbers)
numbers.clear() # Remove everything
print(numbers)
```

```python
# INDEX(VALUE)
numbers=[10,20,30,20,40,50,40,30,20,50,20]
print(numbers.index(20)) # Returns first occurance
# INDEX(VALUE,START)
print(numbers.index(20,4))
# INDEX(VALUE,START,END)
print(numbers.index(40,2,8))
```
