```python
# USER DEFINED FUNCTION
def greet():
    print("Hello")
greet()
```

```python
# FUNCTION WITH PARAMETERS
def start_program():
    print("Program started!")
start_program()
```

```python
# FUNCTION WITH MULTIPLE PARAMETERS
def add(a,b):
    print(a,b)
add(10,20)
```

```python
# FUNCTION WITH RETURN
def mul(x,y):
    return x,y
print(mul(10,20))
```

```python
# FUNCTION IS CALLING ANOTHER FUNCTION
def square(num):
    return num**2
def display_square(val):
    result= square(val)
    print(result)
display_square(4)
```

```python
# MULTIPLE RETURN VALUES
def calculate(a,b):
    total= a+b
    product= a*b
    return total, product
result= calculate(10,20) # It returns "None "
print(result)
result_1, result_2 = result
print(result_1)
print(result_2)
```

```python
# FUNCTION PASS
def auto():
    print("3 Wheels")
def car():
    pass
def cycle():
    print("2 wheels")
auto()
car()
cycle()
```

```python
# DOCSTRINGS
def cube(x):
    '''Return the value of cube'''
    return x*x*x
print(cube(10))
```

```python
# PARAMETERS AND ARGUMENTS
def sum(a,b): # a, b is parameters
    return a+b
print(sum(10,20)) # 10, 20 is arguments
```

```python
# KEYWORD ARGUMENTS
def introduce(name,age):
    print(name,"is",age,"year's old")
introduce(name="William",age=28) # name, age is keyword arguments
```

```python
# POSITIONAL ARGUMENTS
def combine(firstname, lastname):
    return firstname + lastname
print(combine("William ","Sekar")) # William, Sekar  is positional arguments
```

```python
# DEFAULT PARAMETERS
def print_port(port="8080"): # default value is 8080
    print(port)
print_port(1111)
print_port()
print_port(1000)
```

```python
# *ARGS 
def test_1(*args): # Receive in *args parameter 
    print(args) # Print in tuples format
    a,b,c= args # Extracted
    print(a,b,c)
    for i in args: # Using loop
        print(i)
test_1(10,20,30) # Same data type values with positional argumnnts

def test_2(*packets): # Receive in *packets parameter (tuple)
    print(packets) # prin int tuple format
    name, age, status= packets # Extracted
    print(name, age, status)
    for data in packets: # Using loop
        print(data, end=" ") 
test_2("William", 28, True) # different data with positional arguments
```

```python
# KWARGS
def student_data(**kwargs): 
    print(kwargs) # Print in dictionary format
    a,b,c=kwargs # Extracted
    print(a,b,c)
    for data in kwargs:
        print(data)
student_data(name="william", age=28, status=True) # Keyword arguments
```

```python
# POSITIONAL-ONLY PARAMETERS
def t1(a,b,/,c,d): # End with "/"
    print(a,b) # a,b is positional-Only parameters
    print(c,d) # c,d is normal positional parameters
t1(10,20,"Alice",True)
```

```python
# KEYWORDS-ONLY PARAMERTERS
def t2(no,status,*,name,age): # Start with "*"
    print(no,status) # no, status is normal keywords parameters
    print(name,age)  # name, age is keywords-only parameters
t2(no=1,status=True,name="Bob",age=30)
```

```python
# FUNCTINO RETUEN NONE
def t3():
    return
print(t3())
```

```python
# FUNCTION RETURN CONTROL STATEMENTS OR CONDITION STATEMENTS
def t4(num):
    if num>0:
        return "+ive"
    elif num<0:
        return "-ive"
    else:
        return "0's"
print(t4(-45))
print(t4(5.5))
print(t4(0))
    
```
