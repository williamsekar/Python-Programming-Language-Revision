```python
# IMPORT MODULE
import calculator 

print(calculator.add(10, 20))
```

```python
# USING AS FOR SHORT NAME
import calculator as c
print(c.mul(10,5))
```

```python
# IMPORT SPECEFIC METHODS FROM MODULES
from calculator import div,mod
print(div(10,5))
mod(13,2)
```

```python
# IMPORT SPECEFIC METHODS FROM MODULES AND ALSO USING AS
from calculator import div as d,mod as m
print(d(10,5))
m(13,2)
```
