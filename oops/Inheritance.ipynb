```python
# INHERITANCE
class BankAccount: # parent class
    def __init__(self,name,balance):
        self.name = name
        self.__balance = balance
        
    def get_balance(self):
        return self.__balance
        
    def show_balance(self):
        print("Name:",self.name)
        print(f"Balance: Rs.{self.get_balance()}")
        
    def deposit(self,amount):
        if amount > 0:
            self.__balance += amount
            print(f"Rs.{amount} is deposited, balance: Rs.{self.__balance}")
        else:
            print("Invalid amount!")
            
    # normal method      
    def withdraw(self,amount):
        if amount <= 0:
            print("Invalid amount!")
        elif amount > self.__balance:
            print(amount)
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            print(f"Rs.{amount} is withdraw, balance: Rs.{self.get_balance()}")

class SavingAccount(BankAccount):
    def add_interest(self, rate):
        interest = self.get_balance() * rate / 100
        print(f"Saving interest {rate}%: Rs.{interest}")
        super().deposit(interest) # super() -> go to parent class
        
class CurrentAccount(BankAccount):
    def add_interest(self, rate):
        interest = self.get_balance() * rate / 100
        print(f"Current interest {rate}%: Rs.{interest}")
        super().deposit(interest)
        
    # method overriding
    def withdraw(self,amount):
        if amount > 500:
            fee = 5
            total= amount + fee
            super().withdraw(total)
        else:
            super().withdraw(amount)

print("----------------------")
print("Saving account:")
print("----------------------")
t1= SavingAccount("Alice",10000)
t1.show_balance()
t1.deposit(250)
t1.add_interest(5)
t1.withdraw(1500)

print("----------------------")
print("Current account:")
print("----------------------")
t2= CurrentAccount("Bob",1000)
t2.show_balance()
t2.deposit(250)
t2.add_interest(3)
t2.withdraw(600)
t2.withdraw(100)
```
