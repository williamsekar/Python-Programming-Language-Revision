```python
# DUCK TYPING
class Cat:
    def sound(self):
        print("Meow")
class Dog:
    def sound(self):
        print("Bark")
class Cow:
    def sound(self):
        print("Moo")

def make_sound(obj):
    obj.sound()

make_sound(Cat())
make_sound(Dog())
make_sound(Cow())
```

```python
# CLASSIC POLYMORPHISM
class Cat:
    def sound(self):
        print("Meow")
class Dog:
    def sound(self):
        print("Bark")
class Cow:
    def sound(self):
        print("Moo")

animals=[Cat(),Dog(),Cow()]

for animal in animals:
    animal.sound()
```
