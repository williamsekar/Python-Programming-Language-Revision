```python
# METHODS
class Student:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    def display(self):
        print("name: ",self.name)
        print("age: ",self.age)
        print("gender: ",self.gender)
        
student1=Student("william",27,"male")
student2=Student("Prema",28,"female")

student1.display()
student2.display()
```

```python
# CONSTRUCTOR AND METHODS WITH PARAMETERS

class Bankaccount:
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance
    def deposit(self,amount):
        print("Rs.",amount,"deposit successfully")
        self.balance +=amount
    def withdraw(self,amount):
        if self.balance < amount:
            print("Insuffient balance!")
        else:
            self.balance-=amount
            print("Rs.",amount,"withdraw sucessfully")
    def show_balance(self):
        print("Owner: ",self.owner)
        print("Balance:", self.balance)
        
c1=Bankaccount("william",500)
c1.show_balance()
c1.deposit(200)
c1.show_balance()
c1.withdraw(100)
c1.show_balance()
```

```python
# CLASS VARIABLE
class Bankaccount:
    branch="Nellai" # class variable
    def __init__(self,owner,balance):
        self.owner=owner # instance variable or object variable
        self.balance=balance
    def deposit(self,amount): # instalnce method
        print("Rs.",amount,"deposit successfully")
        self.balance +=amount
    def withdraw(self,amount):
        if self.balance < amount:
            print("Insuffient balance!")
        else:
            self.balance-=amount
            print("Rs.",amount,"withdraw sucessfully")
    def show_balance(self):
        print("Owner: ",self.owner)
        print("Balance:", self.balance)
        print("From inner class - Branch: ",Bankaccount.branch) # class name used to get the value
        
c1=Bankaccount("william",500)
c1.show_balance()
c1.deposit(200)
c1.show_balance()
c1.withdraw(100)
c1.show_balance()
print("From Outer class - Branch: ",Bankaccount.branch)
Bankaccount.show_balance(c1) # conceptly mean this
```

```python
# CLASS METHOD
class Bank:
    branch="Nellai"

    @classmethod
    def change_branch(cls,new_branch): # cls - current class, self - current object
        cls.branch=new_branch
# Before
print(Bank.branch)
# after
Bank.change_branch("Chennai")
print(Bank.branch)
```

```python
# STATIC METHOD
class Calculator:
    @staticmethod
    def add(a,b):
        return a+b
print(Calculator.add(10,20)) # class can access the "static method"
t1=Calculator()
print(t1.add(10,20)) # object can access  the "static method"
```
